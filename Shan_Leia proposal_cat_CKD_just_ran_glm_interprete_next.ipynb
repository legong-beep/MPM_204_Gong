```r
# Convert categorical variables to factors so R can compare
cat_df$Breed <- as.factor(cat_df$Breed)
cat_df$Age <- as.factor(cat_df$Age)
cat_df$Sex <- as.factor(cat_df$Sex)
cat_df$Neuter <- as.factor(cat_df$Neuter)
cat_df$Case <- as.factor(cat_df$Case)

# Convert character TRUE/FALSE columns to logical; converts to actual true/false
cat_df$`Heart Murmur` <- as.logical(cat_df$`Heart Murmur`)
cat_df$`Gallop rhythm` <- as.logical(cat_df$`Gallop rhythm`)
cat_df$Arrhythmia <- as.logical(cat_df$Arrhythmia)
```

```r
#verify changes worked before modeling so R doesnt error out due to wrong labeling
str(cat_df)
```

```r
# Check missing values in chosen covariates only
covariates <- c("Periodontal disease", "Age", "Breed", "Sex",
                "Neuter", "Hyperthyroidism", "NSAID treatment",
                "Previous AKI", "Heart Murmur", "Case")

colSums(is.na(cat_df[covariates]))
```

```r
#neuter, heart murmur, and NSAID treatment are missing a alot. Safe to assume NA means "NO NSAID treatment"?
#if so, following code changes that
#heart murmur is being dropped from covariate list due to many missing values
#neuter is dropped too
```

```r
#recoding NA to None for NSAID treatmnet
cat_df$`NSAID treatment`[is.na(cat_df$`NSAID treatment`)] <- "None"
cat_df$`NSAID treatment` <- as.factor(cat_df$`NSAID treatment`)
```

```r
#checking the values from "previous AKI" bc we dont know what categories it contains
table(cat_df$`Previous AKI`)
```

```r
#drop "previous AKI" bc only 10 out of 1,250 cats had previous AKI
```

```r
# Fit a logistic regression model to predict CKD (Case/Control)
# Binary outcome (Case) → binomial family with logit link
# All predictors are fixed effects — no clustering variable exists
# na.omit drops rows with missing values in any covariate
model1 <- glm(Case ~ `Periodontal disease` + Age + Breed + Sex +
                Hyperthyroidism + `NSAID treatment`,
              family = binomial(link = "logit"),
              data = cat_df,
              na.action = na.omit)

summary(model1)
```
